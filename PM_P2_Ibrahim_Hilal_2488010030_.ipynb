```python
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}  # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]   # fitur
y = df['harga']                     # label
```

**REFLEKSI**

1.Apa perbedaan supervised, unsupervised, dan reinforcement learning?

. Perbedaan Supervised, Unsupervised, dan Reinforcement Learning:
   - Supervised: Model belajar dari data yang sudah memiliki label/jawaban target.
   - Unsupervised: Model belajar dari data tanpa label untuk menemukan pola/kelompok tersembunyi.
   - Reinforcement: Model (agent) belajar dari interaksi dengan lingkungan melalui sistem hadiah (reward) dan hukuman (penalty).

2. Mengapa pembagian data latih dan uji penting?

Pentingnya Pembagian Data Latih dan Uji:
   - Untuk menguji kemampuan generalisasi model pada data baru yang belum pernah dilihat sebelumnya, sehingga mencegah terjadinya overfitting.

3. Mengapa alur kerja ML disebut iteratif?

Alur Kerja ML Dikatakan Iteratif:
   - Karena alur pengembangan ML bukan garis lurus serba sekali jadi. Kita perlu berulang kali mengevaluasi hasil, mengubah fitur, menyesuaikan hyperparameter, hingga memperoleh performa model yang optimal
